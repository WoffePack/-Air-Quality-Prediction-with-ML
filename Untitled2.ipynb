```python
from google.colab import files

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
# Baca CSV
df = pd.read_csv('covid_variants_dataset.csv')
```

```python
# Tampilkan informasi awal
display(df.head())
print(df.info())
print("\nJumlah missing values per kolom:")
print(df.isnull().sum())
```

```python
# Drop duplikat
df.drop_duplicates(inplace=True)
```

```python
# Konversi kolom tanggal
if 'Date_Detected' in df.columns:
    df['Date_Detected'] = pd.to_datetime(df['Date_Detected'], errors='coerce')
```

```python
# Drop missing values
df.dropna(inplace=True)
```

```python
# Deteksi anomali
anomaly_condition = (
    (df['Cases'] <= 0) | (df['Cases'] > 1e6) |
    (df['Deaths'] < 0) | (df['Recovered'] < 0) |
    (df['Hospitalized'] < 0) |
    (df['Transmission_Rate'] < 0) | (df['Transmission_Rate'] > 10) |
    (df['Mutation_Count'] < 0) | (df['Mutation_Count'] > 100)
)
anomalies = df[anomaly_condition]
df_cleaned = df.drop(anomalies.index).copy()
```

```python
# Ringkasan
print("\nRingkasan:")
print(f"- Total data awal      : {len(df)}")
print(f"- Jumlah data anomali  : {len(anomalies)}")
print(f"- Jumlah data valid    : {len(df_cleaned)}")
```

```python
# Simpan hasil ke CSV
output_filename = 'covid_variant_dataset_cleaned.csv'
df_cleaned.to_csv(output_filename, index=False)
files.download(output_filename)
```
